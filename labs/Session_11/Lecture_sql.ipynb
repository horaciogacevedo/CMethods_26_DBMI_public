## Relational databases

Roughly speaking, the model of a relational databases is based on a representation of data in a tabular format (like Excel files).

Each row is refer to as *tuple*, a column is refer to as *attribute*, and it has *relational operators* that linked together tables. (figure source wikipedia)

## Structured Query Language (SQL)

SQL is an international standard language for all relational database systems. There are many flavors of it
+ SQLite <https://www.sqlite.org/index.html>. An open-source system implemented as a C library.
+ MySQL <https://www.mysql.com/>. An *openish* source owned by Oracle and aimed to large databases. 
+ MariaDb <https://mariadb.org/>  The open source alternative to MySQL
+ OracleDB <https://www.oracle.com/database/technologies/> It is a commercial solution.

SQL's design goes back to 1970's when computer languages for business tried to mimic human languages (in particular English). So, the *grammar* of the language was expressed in terms of English-language keywords and a syntaxt that combined the keywords into some executable *phrases*. However, SQL is usually case-insensitve despite the convention to capitalize SQL commands.

```bash
SELECT TemperatureC,Pressure
FROM weather
WHERE date== 08/28/2023;
```

It is intuitive that we are selecting from the (table) *weather*, the value of the variables (columns) *TemperatureC* and *Pressure* for the day 08/28/2023. 

From a computer programming perspective, the SQL language is not pretty but its efficiency for large datasets is remarkable.

## SQLite

There are different types of programs that handle SQL databases. Among the mature and powerful database client/servers such as MySQL, MariaDb, OracleDB, Microsoft SQL server, etc. However,  *SQLite is an in-process library that implements a self-contained, serverless, zero-configuration, transactional SQL database engine*

## An example

I will use a very small subsample from the Kaviar database <https://db.systemsbiology.net/kaviar/>

````python
import sqlite3
con = sqlite3.connect("../../Data/Kaviar_random_sample_chr1_hg19.db")
cur = con.cursor()
````

````python
result = cur.execute("SELECT chrom,start,end,vcf_id,in_1kg from variants limit 10")

for row in result:
    print(row)
````

> Don't forget to close the connection once you are done!

````python
con.close()
````

If we are planning to make a new query, we will reinitialize the process

````python
new_con = sqlite3.connect("../../Data/Kaviar_random_sample_chr1_hg19.db")
new_cur = new_con.cursor()
````

````python
new_result = new_cur.execute("SELECT chrom,start,end,vcf_id,in_1kg from variants limit 10")

for row in new_result:
    chrom, start, end, vcf_id, in_1kg = row 
    text_1kg="".join("is in" if in_1kg else "is not in")
    print(f"{chrom}:{start}-{end} variant type {vcf_id} {text_1kg} the 1000Genomes project.")
    
````

## NoSQL databases

This section follows <cite id="64o98"><a href="#zotero%7C11398979%2FXZI34JMR">(Du, Ke-Lin &#38; Swamy, M.  N. S., 2019)</a></cite>

Traditional row-oriented databases have problems with performance particularly with big volumes. Nowadays, big data platforms have expanded into *column-oriented* databases that allows large data compression and very fast query times.

NoSQL database frameworks allow data access based on predifined access primitives such as *key-value* pairs. This data access patterm results in better scalability and performance since they don't required fixed table schemas or support expensive join operations. Nonetheless, NoSQL, does not avoid SQL. 

The value for a given key (or row-ID) can be a collection of couples composed from a name and a value attached to this name. Data access operations, typical CRUD (create, read, update, and delete) operations, have only a key as the address argument. 
For column-NoSQL databases new columns can be added to these collections, without being limited to a rigid scheme.

Popular NoSQL databases include

- MongoDB <https://www.mongodb.com/>
- Apache Cassandra <https://cassandra.apache.org/_/index.html>
- Neo4J <https://neo4j.com/>
- Amazon Dynamo <https://docs.aws.amazon.com/amazondynamodb/latest/developerguide/Introduction.html>
