## Variables in Python <a name="variables"></a>
 
A *variable* in python is a placeholder that will keep some sort of data structure (technically an **object**). 

```python
a=5.0
a+1
```

In these two lines, several things are happening:
+ Python creates the variable *a* and assigns the value 5.0
+ Everytime the variable *a* is invoked, Python replaces the variable by its assigned value.

> Type the following code

```python
5.0=b
```

Note that the computer needs to know that the variable exists before it usses it!

```python
b+1.0
b=2.0
```

Also, remember the conventions for naming variables. Valid variable names 
+ Letters, numbers and underscores
+ Spaces are not allowed
+ Do not use Python **keywords** as variable names
+ Do not use **dots** 


Invalid names for variables include

```python
a.1=5.0
a 1=5.0
print=4.0
```

## Strings in Python <a name="strings"></a>

*Strings* are sequences of alphanumeric characters. Some of these characters can be *invisible* (e.g., spaces, new-lines, etc.)
They start and end with either a single quotation mark or double quotation mark.

```python
my_string1='hi there'
my_string2="hello there"
```

We can access specific locations that are indexed 

> In Python, first elements of a string, list, etc. **start at the zeroth location**

```python
print(my_string1[0])
```

```python
print(my_string1[5])
```

We can also take **slices** or parts of a string

```python
print(my_string2[3:]
```

Or going in the reverse order

```python
print(my_string2[-2])
```

### Operations with strings <a name="stringoperations"></a>

One basic operation with strings is **concatenation**, which means pasting together two strings

```python
my_string3=my_string1+ my_string2
```

Or we can add something else

```python
my_string3=mystring1+":"+my_string2
```

We can also **repeat** concatenating strings

```python
my_string4=3*my_string3
```

### Immutability and Special characters<a name="stringsimmutability"></a>

One may need to change the string

```python
my_string1[1]="o"
```

However, Python consider strings **immutable**, which means that the only way to change a string is to generate a new one with the given changes.

```python
my_string5=my_string1[0]+"o"+my_string1[2:]
```
