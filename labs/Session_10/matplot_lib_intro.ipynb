# Matplotlib

Data summaries and visualizations in Python are normally carried out with Matplotlib (<https://matplotlib.org/>).

Although it is not the most visually appealing library, it normally suffices for most analyses and works very intuitively with Python.
A similar interaction occurs in R with *ggplot2*.

```python
import matplotlib.pyplot as pyplot
# set up data
x=[0,1,2,3,4,5]
y=[0,2,4,14,30,43]

pyplot.ylabel('Speed',fontsize=12)
pyplot.xlabel('Time',fontsize=12)

pyplot.title('Speed vs Time')
pyplot.plot(x,y,'bo-')
pyplot.show()
```

```python
import random 
import matplotlib.pyplot as pyplot
random.seed(50303)

riding_x=[random.randint(-10,10) for i in range(20)]
riding_y=[random.randint(-20,20) for i in range(20)]

swimming_x=[random.randint(-10,10) for i in range(20)]
swimming_y=[random.randint(-20,20) for i in range(20)]

pyplot.scatter(x=riding_x,y=riding_y,c='red',marker='o',label='riding')
pyplot.scatter(x=swimming_x,y=swimming_y,c='green', marker='^',label='swimming')

pyplot.xlabel("Any label")
pyplot.ylabel("Whatever number")
pyplot.title("Random data is fun")
pyplot.legend()
pyplot.show()
```

```python
import random 
import matplotlib.pyplot as pyplot
import numpy as np
random.seed(50303)

riding_x=[random.randint(-10,10) for i in range(20)]
riding_y=[random.randint(-20,20) for i in range(20)]

swimming_x=[random.randint(-10,10) for i in range(20)]
swimming_y=[random.randint(-20,20) for i in range(20)]

pyplot.scatter(x=riding_x,y=riding_y,c='red',marker='o',label='riding')
pyplot.scatter(x=swimming_x,y=swimming_y,c='green', marker='^',label='swimming')

z=np.polyfit(np.array(riding_x),np.array(riding_y),1)
p=np.poly1d(z)

pyplot.plot(riding_x,p(riding_x),'red')

pyplot.xlabel("Any label")
pyplot.ylabel("Whatever number")
pyplot.title("Random data is fun")
pyplot.legend()
pyplot.show()
```

```python
import matplotlib.pyplot as pyplot

labels=('Python','Scala','C#','Java','PHP')
index=(1,2,3,4,5)
sizes=[45,10,15,30,22]

pyplot.bar(index,sizes,tick_label=labels)
pyplot.ylabel('Usage')
pyplot.xlabel('Programming Languages')

pyplot.show()
```

```python
import matplotlib.pyplot as pyplot

labels=('Python','Scala','C#','Java','PHP')
index=(1,2,3,4,5)
#sizes=[45,10,15,30,22]
web_usage=[20,2,5,10,14]
data_science_usage=[15,8,5,15,2]
games_usage=[10,1,5,5,4]

pyplot.bar(index,web_usage,tick_label=labels,label='web')
pyplot.bar(index,data_science_usage,tick_label=labels,label='data science',bottom=web_usage)

web_and_games_usage=[web_usage[i]+data_science_usage[i] for i in range (len(web_usage))]
pyplot.bar(index,games_usage,tick_label=labels,label='games',bottom=web_and_games_usage)

pyplot.ylabel('Usage')
pyplot.xlabel('Programming Languages')
pyplot.legend()
pyplot.show()
```

```python
import matplotlib.pyplot as pyplot
t=range(20)
s=range(30,10,-1)
grid_size='22'
figure=pyplot.figure()

#position=grid_size+'1' # this will be deprecated soon
#position=(2,2,1)
axis1=figure.add_subplot(2,2,1)
axis1.set(title='subplot(2,2,1)')
axis1.plot(t,s)

#position=grid_size+'2'
axis2=figure.add_subplot(2,2,2)
axis2.set(title='subplot(2,2,2)')
axis2.plot(t,s,'r-')

#position=grid_size+'3'
axis3=figure.add_subplot(2,2,3)
axis3.set(title='subplot(2,2,3)')
axis3.plot(t,s,'g-')

#position=grid_size+'4'
axis4=figure.add_subplot(2,2,4)
axis4.set(title='subplot(2,2,4)')
axis4.plot(t,s,'b-')

pyplot.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.ticker import LinearLocator
import numpy as np

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})

# Make data.
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot the surface.
surf = ax.plot_surface(X, Y, Z, cmap=cm.coolwarm,
                       linewidth=0, antialiased=False)

# Customize the z axis.
ax.set_zlim(-1.01, 1.01)
ax.zaxis.set_major_locator(LinearLocator(10))
# A StrMethodFormatter is used automatically
ax.zaxis.set_major_formatter('{x:.02f}')

# Add a color bar which maps values to colors.
fig.colorbar(surf, shrink=0.5, aspect=5)

plt.show()
```
