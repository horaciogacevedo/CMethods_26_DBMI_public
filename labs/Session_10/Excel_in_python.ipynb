## The openpyxl package

A package that allows you to work with Excel files directly is *openpyxl* (<https://openpyxl.readthedocs.io/en/stable/>).

The authors have pointed out some vulnerabilities: by default openpyxl does not guard against quadratic blowup or billion laughs XML attacks (install *defusedxml* to guard against them). Be aware that since Excel (Word, etc.) files are *xml* files
they may be more susceptible to data corruption.

## Example

The following program creates a workbook with two worksheets (it is also available as the Python file `excel_python.py`). We start by creating the workbook and filling the first worksheet.

```python
from openpyxl import Workbook

print("My Excel Example")
workbook = Workbook()
ws = workbook.active
ws.title = 'worksheet_1'
ws.sheet_properties.tabColor = '1072BA'
ws['A1'] = 42
ws['A2'] = 12
ws['A3'] = '=SUM(A1,A2)'
```

Note that `A3` contains a *formula*. openpyxl stores the formula as text; Excel calculates its value when the file is opened.

Now we create a second worksheet and add some values in different ways.

```python
ws2 = workbook.create_sheet(title='worksheet_2')
ws2['A1'] = 3.42
ws2.append([1, 2, 3])
ws2.cell(column=2, row=1, value=15)
```

Finally, we save the workbook into a file.

```python
workbook.save('sample.xlsx')
print("Finishing off")
```

We can read the file back to check what was saved.

```python
from openpyxl import load_workbook

wb = load_workbook('sample.xlsx')
print(wb.sheetnames)
print(wb['worksheet_1']['A3'].value)
```
