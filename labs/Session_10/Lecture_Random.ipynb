Can a deterministic machine produce *random* sequence numbers?

\begin{equation}
\langle U_n\rangle= U_0,U_1,U_2,\ldots
\end{equation}
where $0\le U_n <1$ for any $n$?

In principle, any program in a deterministic machine will produce an output that is entirely predictable. However, in any modern computer language we can generate a *random sequence* that we hope to "behave as if they are truly random".

Don Knuth writes:

" The axioms of probability theory are set up so that abstract probabilities can be computed readily, but nothing is said about what probability really signifies, or how this concept can be applied meaningfully to the actual world... a proper definition of probability depends on obtaining a proper definition of a random sequence."

## Simulation

We call it a *simulation study* if we empirically determine probabilities by means of (computational) experimentation.

The first task of a simulation is to generate values coming from a uniform random variable $X$ in the interval $(0,1)$ (denoted as $U(0,1)$). This means:

If $u,v $ are real numbers $0< u<v <1$, and the random variable $X \sim U(0,1)$, then $P(u\le X <v)= v-u$. For instance, $P(\frac{1}{7}\le X < \frac{2}{7})= \frac{1}{7}$.


To produce such a random variable, our computers produce *pseudo*-random sequence numbers that are "indistinguishable" from a sample taken from $X\sim U(0,1)$. Most random number generators start with an initial value $X_0$ called the **seed**, and then recursively compute values by selecting positive integers $a$, $c$, and $m$ and calculating

\begin{equation*}
X_{n+1} = (a X_n +c )\mod m\quad (n \ge 0)
\end{equation*}

Thus each $X_n$ is either $0,1, \ldots, m-1$ and the quantity $X_n/m$ is taken as an approximation of $X\sim U(0,1)$.

With a proper selection of $a,c, $ and $m$, it can be shown that the sequence of numbers *appears* to be generated from a uniform random distribution $U(0,1)$.

At the minimum, we expect that a random number generator produces different sequences which are statistically uncorrelated to each other.


The basic module *random* generates uniform random numbers, but it is possible to use *numpy*. (Both use the Mersenne Twister generator, which is much better than the simple recursion above.)

```python
import random
random.seed(2026)
print("Uniformly Distributed Random numbers between (0,1) ")
print("%.4f"%random.uniform(0,1))
print("%.4f"%random.uniform(0,1))
```

```python
import numpy as np
np.random.seed(2026)
print("Uniformly Distributed Random numbers between (0,1) ")
print(f"{np.random.uniform(0,1):.4f}")
print(f"{np.random.uniform(0,1):.4f}")
```

Let's verify empirically, how accurate are our random number generators

```python
import math
my_mean=0; my_stdev=1.0; n=50000
u=np.random.normal(my_mean,my_stdev,n)
m=sum(u)/n; s =math.sqrt(sum((u-m)**2)/(n-1))
print(f"Mean {my_mean:.2f} and Empirical mean {m:.4f}")
print(f"Standard Deviation {my_stdev:.5f} and Empirical Stdev {s:.5f}")
```

## Generating a Normal Distribution

Based on probability theory, it can be shown that if we have two INDEPENDENT random variables $X_1, X_2 \sim U(0,1)$ we can generate two independent standard normal distribution $N(0,1)$. This means normally distributed with zero mean and standard deviation equals 1 (this is known as the Box-Muller transform).

Then, we define

\begin{align*}
Y_1 &= \sqrt{-2 \log X_1} \cos (2\pi X_2)\\
Y_2 &= \sqrt{-2 \log X_1} \sin (2\pi X_2)
\end{align*}

It can be theoretically verified that $Y_1$ and $Y_2$ are INDEPENDENT and both have $N(0,1)$.
