## The Problem

As we have seen in class, there are a number of issues between software versioning. This problem scalates

+ Development normally takes place on a local machine
+ How a locally developed solution will behave in another environment?
+ How to preserve the software with minimal modifications?
+ What if you want to take your software solution to scale?

> In what follows we will show the most common solution **Docker** (www.docker.com)


## Solution: Containers

A software container is an abstraction that packages applications and dependencies so they run reliably across computing environments.

+ A **container image** it is a software package that contains all your specific requirements.
+ These images are single package that can be deployed anywhere provided that you have a basic infrastructure.
+ It can run in multiple software platforms

## Container Definition File

It is a text file that describes how to build a container image, including packages, environment variables, metadata, and startup behavior.

```dockerfile
FROM ubuntu:24.04
RUN apt-get update && apt-get install -y python3 python3-pip vim wget
WORKDIR /workspace
CMD ["/bin/bash"]
```



# Container Process

## Advantages of Containers

+ Near-native performance. There is no much overhead for the use of a container in comparison to native applications.
+ Reproducibility. Everything is packaged according to a fix recipe, and the individual components can be verified.
+ Portability. Once they are built, they can deploy almost in any other system.
+ Software freedom. You can bring your own software.


## Limitations of Containers

- Architecture dependent in the sense that they are limited to CPU architecture (x86_64, ARM)
- Kernel compatibility required between the host and container. 
- Filesystem isolation considerations as paths inside and out of the container are different.

## From Singularity to Apptainer

The development of Singularity (https://sylabs.io/singularity/) resolved some of the security issues and was widely addopted in HPC systems. 

The Singularity community project evolved into Apptainer (https://casrai.org/compare/apptainer-vs-singularity). Most workflows, formats, and commands remain compatible.

<img src="/fig_singularity.png" width="200"/>


## Apptainer: A Container Platform for HPC

- Reproducible and portable
- No privileged daemon
- Suitable for shared HPC systems
- https://apptainer.org

## Features of Apptainer

- Single SIF image
- No root escalation
- Same user identity inside/outside container
- Multi-tenant support

## GPU-Accelerated Apptainer Containers

Use --nv to provide access to host GPU drivers and libraries.

## Most Common Apptainer Use Cases

- Newer software stacks
- Commercial binaries
- Conversion from Docker images

## Dockerfile Example: Scientific Python

```dockerfile
FROM python:3.12
RUN pip install numpy scipy pandas matplotlib jupyter
WORKDIR /notebooks
EXPOSE 8888
CMD ["jupyter","lab","--ip=0.0.0.0","--allow-root"]
```

## Dockerfile Example: R

```dockerfile
FROM rocker/r-base:latest
RUN R -e "install.packages(c('tidyverse','ggplot2'))"
CMD ["R"]
```

# Apptainer Definition File

```text
Bootstrap: docker
From: ubuntu:24.04
%post
    apt-get update
    apt-get install -y python3 python3-pip
%runscript
    exec python3 "$@"
```
