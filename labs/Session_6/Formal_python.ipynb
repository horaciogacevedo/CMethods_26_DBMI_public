## Functions 
The normal way to define a function is to use the keyword *def* 

```python
def floatrange(start,stop,inc):
    result=[]
    while start<stop:
        result.append(start)
        start +=inc # start = start + inc
    return result
```

```python
print(floatrange(0.5,4,1))
```

### Generator Functions

This type of functions do not have **return** statement, but instead they use the **yield** statement.


```python
def floatrange2(start,stop,inc):
    while start<stop:
        yield start
        start +=inc
```

```python
b=floatrange2(0.5,4,1)
print(b.__next__())
print(b.__next__())
print(b.__next__())
print(b.__next__())
print(b.__next__())
```

## Exceptions

An **exception** is an object that is *raised* under specific circumstances. Upon an exception call the program gets redirected towards the exception handler following backwards the function evaluation stack. If it finds a **finally** block it is executed and the search for the exception handler resumes.

We can think of the **finally** block as a cleaner function.  

```python
import sys

def openmyfile(myfile):
    try: 
        try:
            mf=open(myfile,r)
        except IOError as e:
            print("Trouble, trouble!"% e)
    finally:
        if mf:
            mf.close()
```

## Inheritance 

In Python inheritance is used mostly to reuse code, because if a method from a base class is change, it automatically gets deployed to all inherited classes.


```python
class NFLPlayer(object):
    def __init__(self, player,position, team):
        self.__player=player
        self.__position=position
        self.__team=team
    
    def getPlayer(self):
        return self.__player
    
    def setPlayer(self,player):
        self.__player=player
        
    def getPosition(self):
        return self.__position
    
    def setPosition(self,position):
        self.__position=position
    
    def getTeam(self):
        return self.__team
    
    def setTeam(self,team):
        self.__team=team
    
    def __str__(self):
        if self.__team is not None:
            return "%s plays as %s for %s"%(self.__player,self.__position,self.__team)
        else:
            return "%s plays as %s and is a free-agent"%(self.__player,self.__position)

class Recruit(NFLPlayer):
    def __init__(self,player,position, team=None):
        super(Recruit,self).__init__(player,position,team)
    
class Professional(NFLPlayer):
    def __init__(self,player,position,team,salary):
        super(Professional,self).__init__(player,position,team)
        self.__salary=salary
        
    def __str__(self):
        bigsalary=1000000 
        if self.__salary> bigsalary:
            return "%s and earns big bucks!"% super(Professional,self).__str__()
        else:
            return "%s and barely makes it!"% super(Professional,self).__str__()
    
```

## Testing  your code

Python has several ways to perform *unit testing*. We will review the simplest of them and later on we will implement more sophisticated ones.


```python
import math
def isprime(n):
    """ This is a silly example
    >>> isprime(0)
    False
    >>> isprime(1)
    False
    >>> isprime(2)
    True
    >>> isprime(3)
    True
    >>> isprime(4)
    False
    >>> isprime(5)
    True
    """
    if n <= 0 or n==1 :
        return False
    if n==2 :
        return True
    else:
        for x in range(2,int(math.sqrt(n))+1):
            if n%x ==0:
                return False
        return True
                       
def _test():
    import doctest
    doctest.testmod()

if __name__=="__main__":
    _test()
           
```
