## Unix <a name="unix"></a>

### A Brief History of Unix <a name="a-brief-history-of-unix"></a>

In the beginning...

Unix is an operative system that has been trademarked by several companies, including AT&T, Novell, etc.
It was created in 1969 by a group of scientist at AT&T's Bell Labs with the objective to create an operative system
that was both a multiuser and multitasking. The computer scientist Bill Joy is considered as the primary designer of the Berkeley
Software Distribution (BSD) version of Unix with considerable contributions from Ken Thomson.

![Bill Joy](../../Figures/Bill_Joy.jpg)
![Ken Thompson and Dennis Richie](../../Figures/Ken_Thompson_and_Dennis_Ritchie.jpg)
(Source: wikipedia)

The *C programming language* was created by Dennis M. Ritchie and Ken Thomson, and it is an integral part of Unix. As it turns out, the whole Unix was rewritten in C language. This transition made Unix *portable* since it did not depend any longer on the assembly language for a particular machine. It was just needed a C compiler and viola you will have Unix!

> *Fun Fact:* Unix was not named until about a year into its development and was preceded by *Multics* project. Allegedly, Thomson suggested the name *Unics* (Uniplexed Information and Computing Service, pronounced as "eunuchs"), and Ritchie changed the spelling to *x*.

The first disjunction of Unix occurred when the Berkely Software Distribution (BSD)
split from the AT&T system V.

## Kernel and Shell <a name="kernel-and-shell"></a>

The **kernel** is the core of the operative system. The system's bootstrap program (normally a tiny program code) loads the kernel into memory at startup. The kernel consists of a set of routines, mostly written in C, that establish the communication with the hardware.
The kernel manages the system's memory, schedules processes, and performs other tasks in the background. Thus, even when you don't *see* the kernel iw constantly working.

Computers do not have any inherent capability of translating user commands into action. That requires an interpreter, called the **shell**. This is the interface between the user and the kernel. Once you type something in the keyboard the shell inspects it for special characters or commands and it communicates the result to the kernel for processing.

> The kernel is represented by the file */boot/vmlinuz*. The shells are stored in */bin* and */usr/bin*

**Files** are containers for storing static information. Directories and devices (e.g. usb drive) are considered *files*.

**Processes** are programs in execution. You can think of processes as *living* entities that are created, destroyed, or changed in priorities from the foreground to the background or viceversa.

**System calls** are special programs that access kernel services by invoking them.

> A C program in windows cannot access directly system calls but through the standard library. Unix programmers can use the standard library or the system call directly.

## Unix File System <a name="unix-file-system"></a>

### Files and filenames <a name="files-and-filenames"></a>

A **file** in Unix is a container for storing information. There are three main types of files

- Ordinary file (or regular file). It contains a stream of data.
  - A *text file* contains only printable characters from the ASCII table.
  - A *binary file* contains printable and nonprintable characters from the entire ASCII range.
- Directory file. Normally called a *Folder* in Windows and MacOsX, and it contains an entry for every file and subdirectory that it houses. Each entry has two components
  - The filename
  - A unique identification number for the file or directory (called the *inode*)

- Device file. This represents a device or peripheral (e.g. USB drive), and *it doesn't contain anything*.

> *File attributes* Each file has attributes that helps the kernel decide what to do with the device file. Those attributes are stored somewhere else.

![Ascii table](../../Figures/ASCII-Table-wide.svg)
(wikipedia)

**Filenames** can be up to 255 characters long. The **extension** is optional, and whereas there are few restrictions on the names some best practices suggest that

- It should not start with / or with the NULL character
- Avoid unprintable characters
- Use alphanumeric characters
- Use `.` or `_` or `-` are encouraged instead of BLANK SPACES
- Do not start a filename with a "-"

## Basic Bash Programming <a name="basic-bash-programming"></a>

Bash is the shell, or command language interpreter, for the GNU-Linux operative system. It is a programming language in of its own. The name comes for the acronym *Bourne-Again SHell*, since Stephen Bourne created the Unix shell sh.

There are several ways to run a shell script

````bash
chmod +x script_file
./script_file
````

Another way is to use a direct call to the shell

````bash
/bin/sh script_file
````

Yet another is to add the following line at the beginning of your script to force the current shell to execute the script using Bash

````bash
#!/bin/sh
````

Shell variables can be either *shell environment variables* or *user-defined environment variables*. For instance `HOME` is an environment variable, whereas something like `MYOWNFILE` would be a user-defined environment variable. There are also *read-only* environment variables in Bash such as `$0` (name of the program), `$?` (exit status of the most recent command.

A simple example using `$` to read a variable

````bash
name="Horacio Gomez"
echo "my name is $name"
````

**Sample output:**
```
$ name="Horacio Gomez"
$ echo "my name is $name"
my name is Horacio Gomez
```

Note that when a variable is created (e.g., `name` in the previous chunk of code), it is not automatically known to subsequent command or shell calls.

Unix has a powerful way to process outputs from programs into an input for another program (*pipes*).

### Redirection and Pipes <a name="redirection-and-pipes"></a>

In Unix, the term *terminal* represents either, screen, display, or keyboard! Streams of characters (i.e., sequence of bytes) are associated with a default device. So, we have that

- **Standard Input** is the file or stream representing the input. This can come from the keyboard, a file or from the output of another command or shell.
- **Standard Output** is the file or stream represented output. This can be sent to the keyboard, a file, a terminal (default)
- **Standard Error** is the file or stream representing error messages that emante from the command or shell.

For instance

````bash
head Darwin_origin_of_species.txt > toplines.txt
````

The output of the `head` command applied to the file *Darwin_origin_of_species.txt* is sent or redirected (`>`) to the file *toplines.txt*.
Then, we can see that it is the case

````bash
cat toplines.txt
````

**Sample output:**
```
$ cat toplines.txt
The Project Gutenberg EBook of On the Origin of Species, by Charles Darwin

       1st Edition
```

To direct the standard error, we have a nice (and sometimes cryptic) mechanism

````bash
cat I_forgot_the_file 2> errorfile
cat errorfile
````

**Sample output:**
```
$ cat I_forgot_the_file 2> errorfile
$ cat errorfile

```

We can combined streams into a single one stream as follows

- `1>&2` Sends the standard output to the destination of the standard error
- `2>&1` Sends the standard error to the destination of the standard output

Let's check this simple code

````bash
echo "Where is the file" 2> error.txt 1>&2
cat error.txt
````

**Sample output:**
```
$ echo "Where is the file" 2> error.txt 1>&2
$ cat error.txt
Where is the file
```

### More on variables and shell commands <a name="more-on-variables-and-shell-commands"></a>

The command `set` gives positional values to the argument provided.

````bash
date
set `date`
echo "$@"
echo "$2 $3, $6"
````

**Sample output:**
```
$ date
Tue Sep  1 09:18:52 CDT 2020
$ set `date`
$ echo "$@"
Tue Sep  1 09:18:52 CDT 2020
$ echo "$2 $3, $6"
Sep 2, 2020
```

The command `set` takes the output of the command `date` and gives positional arguments (values) to each entry (in this case denoted by $2, $3, and $6).

### Program control flow <a name="program-control-flow"></a>

*If-then-elif-else-if* statement

````bash
if expression
  then
    then-commands
fi
````

*test* evaluates and expression and returns true or false status.

````bash
test expression
````

Check the file *if_demo.sh*

````bash
#!/bin/sh
if test $# -eq 0
	then
		echo "Usage: $0 file"
		exit 1
	elif test $# -gt 1
		then
			echo "Usage: $0 file"
			exit 1
	else
	ls "$1" 2> /dev/null 1>&2
	if test $? -ne 0
		then
			echo "$1: not found in the present directory"
			exit 1
	fi
	if test -f "$1"
		then
			filename="$1"
			set `ls -il $filename`
			inode="$1"
			size="$6"
			echo "Name\tInode\tSize"
			echo
			echo "$filename\t$inode\t$size"
			exit 0
		elif test -d "$1"
			then
				nfiles=`ls "$1" |wc -w `
				echo "The number of files in the directory is $nfiles"
				exit 0
		else
			echo "$0: argument must be an ordinary file"
			exit 1
	fi
fi
````

### For statement <a name="for-statement"></a>

The syntax for this statement is

````bash
for variable [in argument list]
do
  command list
done
````

Let's explore this with an example

````bash
#!/bin/bash
for i in {1..5}
do
    echo "print $i"
done
echo "More for"
for ((k=1;k<=10;k++))
do
	echo "Welcome $k times"
done
````

### While loop <a name="while-loop"></a>

The syntax is

````bash
while expression
do
    command-list
done
````

Here is a silly example of usage

````bash
#!/bin/bash
secretagent=Bourne
attempts=10
output=yes
echo "Who is knocking? "
echo "Enter your codename"
read codename
i=0
while test "$codename" != "$secretagent"
do
	echo "Code is not right. Try again!"
	read codename
	((i=i+1))
	if test $i -gt $attempts
		then
			output=no
			break
	fi
done
if test $output == "yes"
	then
		echo "Access granted"
	else
		echo "Acces denied"
fi
exit 0
````

### Until loop <a name="until-loop"></a>

It is pretty similar to while, but we don't have it in python.

````bash
until expression
do
   command-list
done
````

Here is the version with *until* loop of the same silly example

````bash
#!/bin/bash
secretagent=Bourne
attempts=10
output=yes
echo "Who is knocking? "
echo "Enter your codename"
read codename
i=0
until test "$codename" == "$secretagent"
do
	echo "Code is not right. Try again!"
	read codename
	((i=i+1))
	if test $i -gt $attempts
		then
			output=no
			break
	fi
done
if test $output == "yes"
	then
		echo "Access granted"
	else
		echo "Acces denied"
fi
exit 0
````

### Case statement <a name="case-statement"></a>

The `case` statement provides a mechanism of multiway branching similar to a nested `if` statement. This command is really missed in Python.

````bash
#!/bin/bash
echo "Select from the following icecream options"
echo " v:	Vanilla"
echo " c:	Chololate"
echo " s:	Strawberry"
echo " n:	Neopolitano"
echo " q:	Quit and select other"
echo "Enter your option and press <Enter>: \c"
read option
case "$option" in
	v)	echo "Plain"
		;;
	c)	echo "Right choice!"
		;;
	s)	echo "Ok choice"
		;;
	n) 	echo "You couldn't make your mind could you?"
		;;
	q)	echo "Better go for Ben and Jerry's"
		exit 0
		;;
esac
exit 0
````
