### Hint

Python programs can receive values from the command line through the `sys` module. The list `sys.argv` holds them:

+ `sys.argv[0]` is the name of the script itself (the very first argument)
+ `sys.argv[1]`, `sys.argv[2]`, ... are the values typed after the script name

Below is an example script, `basic_input.py`. Run the next cell to save it to a file, and the one after to run it with two arguments.

```python
%%writefile basic_input.py
import sys

argument1=sys.argv[1]
argument2=sys.argv[2]
veryfirst=sys.argv[0]

print(f"The first argument was {argument1}")
print(f"and the second argument was {argument2}")
print(f"and the very first argument was {veryfirst}")
```

```python
!python basic_input.py 10 hello
```

Note that every argument arrives as a **string**, even if it looks like a number. To do arithmetic you must convert it (for example with `float()`), and that conversion fails when the text is not a number. Think about how you can use that to check the input.
