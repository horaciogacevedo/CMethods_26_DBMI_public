On a Mac you use the **Terminal** app (Applications > Utilities > Terminal, or press `Cmd + Space` and type *Terminal*). It runs the same kind of *bash/zsh* shell as Git Bash on Windows, so the git commands are identical.

## Mac Setup

### Git setup

Check whether git is already installed. In Terminal type

```bash
git --version
```

If git is missing, macOS will offer to install the **Command Line Developer Tools**. Accept, or run

```bash
xcode-select --install
```

Alternatively, if you use [Homebrew](https://brew.sh): `brew install git`.

Tell git who you are (once):

```bash
git config --global user.name "Your Name"
git config --global user.email "your_email@example.com"
```

### Creating the SSH key pair

```bash
mkdir -p ~/.ssh
cd ~/.ssh
ssh-keygen -t ed25519 -C "your_email@example.com"
```

Accept the default file name and (recommended) set a passphrase. It will create

1. `id_ed25519` (your private key, **do not share**)
2. `id_ed25519.pub` (this one you will share)

Copy the public key directly to the clipboard with

```bash
pbcopy < ~/.ssh/id_ed25519.pub
```

or show it with `cat ~/.ssh/id_ed25519.pub` and copy everything from *ssh-ed25519* until the very end (use a plain text editor such as TextEdit in plain-text mode, not Microsoft Word).

### Add the key to the ssh-agent (and the macOS Keychain)

Start the agent

```bash
eval "$(ssh-agent -s)"
```

Create (or edit) the file `~/.ssh/config` so macOS remembers your key between restarts

```bash
touch ~/.ssh/config
open -e ~/.ssh/config
```

and add these lines

```
Host github.com
  AddKeysToAgent yes
  UseKeychain yes
  IdentityFile ~/.ssh/id_ed25519
```

Then add the key

```bash
ssh-add --apple-use-keychain ~/.ssh/id_ed25519
```

> Note. On macOS older than 12 (Monterey) the option is `-K` instead of `--apple-use-keychain`.

Thanks to the Keychain, the Mac (unlike Windows) will normally **not** forget the key after a restart.

### Testing the connection with GitHub

```bash
$ ssh -T git@github.com
Hi your-github-username! You've successfully authenticated, but GitHub does not provide shell access.
```

The first time, you will be asked whether you trust the host `github.com`. Type `yes`.

## Cloning the class repository

Choose a location for your repositories, for example `~/Documents/GitHub_repos/`

```bash
mkdir -p ~/Documents/GitHub_repos
cd ~/Documents/GitHub_repos
```

In GitHub, go to the [class repository](https://github.com/horaciogacevedo/CMethods_26_DBMI_public), click the green **Code** button and be sure that the **SSH** tab is selected. Copy the address.

<img src="../../Figures/fig_github_cloning.png" width="400"/>

We use Git Bash once again and paste it

```bash
git clone git@github.com:OWNER/REPOSITORY.git
```

(Replace `OWNER/REPOSITORY` with the actual GitHub account and repository name, for example `horaciogacevedo/CMethods_26_DBMI_public.git`.) 

You can then see the repository in **Finder**, or go through the Terminal.

## Updating your repository

To see the status of the current repository is

```bash
git status
```

Make **pulls** often to keep up with new material.

```bash
cd ~/Documents/GitHub_repos/REPOSITORY
git pull origin main
```

The current branch name (`main`, or `master` in older repositories) can be checked with `git branch`. If your repository uses `master`, use `git pull origin master`.

> The operation *pull* is not automatic. You should invoke this command as needed.

## Problems with a new pull

Running even a single cell of a jupyter notebook alters the file, so a later pull may fail because your local copy differs from the repository.

+ Do not panic
+ (Optional) copy the notebook you modified to another name if you want to keep your changes
+ Discard the local changes

```bash
git checkout -- .
```

(or `git restore .`)

+ Do a new pull

```bash
git pull origin main
```

## Troubleshoot

+ If the pull asks for a passphrase every time, or fails with `Permission denied (publickey)`, reload the key:

```bash
eval "$(ssh-agent -s)"; ssh-add --apple-use-keychain ~/.ssh/id_ed25519; ssh -T git@github.com
```

+ `Repository not found`: check the repository address and that you have been given access.
+ Since macOS Catalina the default shell is *zsh*; every command above works the same in zsh and bash.
+ Jupyter tip: if you launch Jupyter from Terminal, remember to `cd` into the repository folder first.
