### Creating the SSH key pair

Open Git Bash. If you are using an UAMS computer substitute your username below (mine is *gomezacevedohoracio*) and change the location or username accordingly.
Note that the email that you use is the one that you use to create your account in Github.

```bash
cd /c/Users/gomezacevedohoracio/
mkdir -p .ssh
cd .ssh
ssh-keygen -t ed25519 -C "your_email@example.com"
```

If you accept the defaults, it will create a couple of files

1. `id_ed25519` (your private key, **do not share**)
2. `id_ed25519.pub` (this one you will share)

> Note. Older tutorials use `ssh-keygen` with RSA (`id_rsa`). That still works with GitHub, but ed25519 is the current recommendation. If you use RSA, replace `id_ed25519` by `id_rsa` everywhere below.

You can use the *cat* command to show the public key

```bash
$ cat id_ed25519.pub
ssh-ed25519 AAAAC3NzaC1lZDI1NTE5AAAAI... your_email@example.com
```

I cut the output, but you need to copy (in [Notepad++](https://notepad-plus-plus.org/)) for instance, not Microsoft Word) everything starting from *ssh-ed25519* until the very end.

Next, you need to **activate** the *ssh-agent* as follows

```bash
$ eval $(ssh-agent)
```

and add your key to the *ssh-agent*

```bash
$ ssh-add ~/.ssh/id_ed25519
Identity added: /c/Users/gomezacevedohoracio/.ssh/id_ed25519 (your_email@example.com)
```

> Note. Be sure to use the path where your key is actually located.

Invoke the following command

```bash
$ ssh -T git@github.com
Hi your-github-username! You've successfully authenticated, but GitHub does not provide shell access.
```

The first time, you will be asked whether you trust the host `github.com`. Type `yes`.

## Updating your repository

The repository will keep on growing, so the best way to keep up is to make **pulls** often.

In Git Bash move to the repository location

```bash
cd /c/Users/gomezacevedohoracio/Documents/GitHub_repos/REPOSITORY/
```

> Note that once you move to that directory, the branch name (`main`, or `master` in older repositories) appears. This is an indication that you have a git repository.

<img src="../../Figures/fig_bashgit_master.png" width="600"/>


To see the status of the current repository is

```bash
git status
```
This is the output that I got while I was working on the lecture (it is not a windows computer)

<img src="../../Figures/fig_repository_checking.png" width="600"/>

Finally, you just update it with a pull

```bash
git pull origin main
```

> The operation *pull* is not automatic. You should invoke this command as needed.
>
> If your repository still uses `master` as its default branch, use `git pull origin master`. You can check the branch name with `git branch`.

## Problems with a new pull

Due to the nature of jupyter notebooks, if you run even a single cell, the notebook will be automatically *altered*, and if you save it, that will keep a local copy that differs from the repository. Thus, you might get an error trying to pull again.

+ Do not panic if you get an error when you try to pull
+ Discard your local changes using the checkout command

```bash
git checkout -- .
```

(In recent versions of git you can also use `git restore .`)

+ Do a new pull

```bash
git pull origin main
```

> Tip. If you want to keep your changes (for example, your own notes), make a copy of the notebook with a different name **before** discarding changes.

## Troubleshoot

I have found something annoying in Windows: it tends to forget the keys added. If the pull does not work, proceed as follows.

Open Git Bash and use the following command

```bash
eval $(ssh-agent);ssh-add ~/.ssh/id_ed25519 ; ssh -T git@github.com
```

Then, you can proceed to update your repository or make a new **pull**.

Other common errors:

+ `Permission denied (publickey)`: the key was not added to the agent, or the public key was not added to your GitHub account. Repeat the `ssh-add` step and check **Settings > SSH and GPG keys**.
+ `Repository not found`: check the repository address and that you have been given access to it.
